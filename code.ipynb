```python
# =========================================================
# Simple Qubit Simulation
# =========================================================
#
# Zwei getrennte Wege, je nach Ziel:
#   1) Statevector-Analyse  -> für Bloch-Sphäre (plot_bloch_multivector)
#   2) Messung / Sampling   -> für Histogramm (plot_histogram)
#
# Wichtig: plot_bloch_multivector() erwartet einen Statevector,
# KEIN counts-Dictionary aus einer Messung. Ein gemessenes Qubit
# ist kollabiert und hat keine Bloch-Vektor-Information mehr.

from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator, Aer
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram, plot_bloch_multivector, array_to_latex
```

### Simple Qubit Simulation

- when we increase the amount of times the simulation goes through (shots = _ ), we still get fast results. That is because our quantum operation with circuit.h(qr) is fairly simple. More complex quatum operations &rarr; longer calculation time 

```python


# ---------------------------------------------------------
# 1) Circuit ohne Messung -> für Bloch-Sphären-Darstellung
# ---------------------------------------------------------

qc_state = QuantumCircuit(3)
qc_state.h([0, 1, 2])
qc_state.s(1)
qc_state.sdg(2)
qc_state.draw(output='mpl', scale=0.8)

# Statevector direkt aus dem Circuit berechnen (keine Simulation nötig)
state = Statevector.from_instruction(qc_state)

# Optional: den Zustand als Vektor/Matrix in LaTeX anzeigen
array_to_latex(state, prefix="|\\psi\\rangle = ")

# Bloch-Sphären für alle 3 Qubits anzeigen
plot_bloch_multivector(state)


# ---------------------------------------------------------
# 2) Circuit mit Messung -> für Histogramm der Ergebnisse
# ---------------------------------------------------------

qr = QuantumRegister(3, 'q')
cr = ClassicalRegister(3, 'c')
qc_measure = QuantumCircuit(qr, cr)

qc_measure.h([0, 1, 2])
qc_measure.s(1)
qc_measure.sdg(2)
qc_measure.measure(qr, cr)
qc_measure.draw(output='mpl', scale=0.8)

simulator = AerSimulator()
result = transpile(qc_measure, backend=simulator)
counts = simulator.run(result, shots=1024).result().get_counts()

plot_histogram(counts)
```
