### Simple Qubit Simulation

- when we increase the amount of times the simulation goes through (shots = _ ), we still get fast results. That is because our quantum operation with circuit.h(qr) is fairly simple. More complex quatum operations &rarr; longer calculation time 

```python
circuit = QuantumCircuit(3)
circuit.h([0, 1, 2])
circuit.s(1)
circuit.sdg(2)
circuit.draw(output='mpl', scale = 0.8)
```

```python
simulator = Aer.get_backend('qasm_simulator')
result = transpile(circuit, backend = simulator).result()
counts = result.get_counts
```
